# Анализ продаж электронных устройств

**Тема:** Продажи электроники в магазине  


### Столбцы DataFrame:
| Столбец | Описание |
|---|---|
| `Категория` | Основная категория товара (Смартфон, Ноутбук, Планшет) |
| `Модель` | Конкретная модель устройства |
| `Цена_руб` | Цена в рублях |
| `Количество` | Количество проданных единиц |

```python
# Импорт библиотеки pandas
import sys
!{sys.executable} -m pip install pandas
import pandas as pd
```

---
## 1. Создание исходного DataFrame

```python
# Словарь с данными о продажах электроники
data = {
    "Категория": ["Смартфон", "Смартфон", "Ноутбук", "Ноутбук", "Планшет",
                  "Планшет", "Смартфон", "Ноутбук", "Планшет", "Смартфон"],
    "Модель":    ["iPhone 15", "Samsung S24", "MacBook Air", "Lenovo IdeaPad",
                  "iPad Air", "Samsung Tab S9", "Xiaomi 14", "ASUS VivoBook",
                  "Huawei MatePad", "OnePlus 12"],
    "Цена_руб":  [89990, 79990, 129990, 54990, 69990, 59990, 59990, 64990, 34990, 54990],
    "Количество":[120, 95, 60, 110, 80, 75, 130, 90, 100, 115],
}

# Создание DataFrame из словаря
df = pd.DataFrame(data)

# Вывод исходной таблицы
print("=== ИСХОДНЫЙ DATAFRAME ===")
print(df.to_string(index=False))
```

---
## 2. Фильтрация данных по показателю (>60.000 руб)

```python
# Булева маска: True для строк, где Цена_руб > 60000
filtered_df = df[df["Цена_руб"] > 60000].reset_index(drop=True)

print("=== ФИЛЬТР: Цена > 60 000 руб ===")
print(filtered_df.to_string(index=False))
```

---
## 3. Добавление новых данных в DataFrame

```python
# DataFrame с новыми позициями
new_rows = pd.DataFrame({
    "Категория": ["Смартфон",       "Ноутбук"],
    "Модель":    ["Google Pixel 8", "HP Pavilion"],
    "Цена_руб":  [69990,             74990],
    "Количество":[88,                72],
})

# Объединяем исходный DataFrame с новыми строками
df = pd.concat([df, new_rows], ignore_index=True)

print("=== ПОСЛЕ ДОБАВЛЕНИЯ НОВЫХ СТРОК ===")
print(df.to_string(index=False))
```

---
## 4. Расчёт общего показателя — суммарная выручка

```python
# Добавляем вычисляемый столбец: выручка = цена × количество
df["Выручка_руб"] = df["Цена_руб"] * df["Количество"]

# Суммируем выручку по всем позициям
total_revenue = df["Выручка_руб"].sum()

print("=== ОБЩАЯ ВЫРУЧКА ===")
print(f"  {total_revenue:,.0f} руб.")
```

---
## 5. Вывод результатов: фильтр, показатель, группировка

```python
# 5а. Отфильтрованный DataFrame (с учётом нового столбца Выручка_руб)
filtered_df = df[df["Цена_руб"] > 60000].reset_index(drop=True)
print("=== ОТФИЛЬТРОВАННЫЙ DATAFRAME (Цена > 60 000 руб) ===")
print(filtered_df.to_string(index=False))

# 5б. Общая выручка
print(f"\n=== ОБЩАЯ ВЫРУЧКА ПО ВСЕМ ДАННЫМ ===")
print(f"  {total_revenue:,.0f} руб.")

# 5в. Группировка по категории: количество позиций, продажи, средняя цена, выручка
grouped = (
    df.groupby("Категория")
      .agg(
          Позиций      =("Модель",       "count"),
          Продано_шт   =("Количество",   "sum"),
          Средняя_цена =("Цена_руб",     "mean"),
          Выручка_руб  =("Выручка_руб",  "sum"),
      )
      .round(0)
      .sort_values("Выручка_руб", ascending=False)
)
print("\n=== ГРУППИРОВКА ПО КАТЕГОРИИ ===")
print(grouped.to_string())
```

---
## 6. Сохранение DataFrame в CSV-файл

```python
df.to_csv("electronics_sales.csv", index=False, encoding="utf-8-sig")

print("Файл успешно сохранён: electronics_sales.csv")
print(f"   Строк: {len(df)} | Столбцов: {len(df.columns)}")
```
